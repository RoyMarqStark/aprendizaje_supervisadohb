# 1. Limpieza de Datos

```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv('data/titanic.csv')
```

```python
df.head()
```

```python
print("Valores nulos por columna")
print(df.isnull().sum())
```

## 1.1 Tratamiento de valores nulos

```python
# Hacemos una copia para el dataset limpio
df_clean = df.copy()

# Imputamos la edad con la mediana
df_clean['Age'] = df_clean['Age'].fillna(df_clean['Age'].median())

# Imputamos Embarked con la moda (el valor más común)
df_clean['Embarked'] = df_clean['Embarked'].fillna(df_clean['Embarked'].mode()[0])

# Para Cabin, dado que tiene muchos valores nulos, llenamos con 'Unknown'
df_clean['Cabin'] = df_clean['Cabin'].fillna('Unknown')

# Verificamos que no queden valores nulos
print("Valores nulos en df_clean:")
print(df_clean.isnull().sum())
```

```python
df_clean.head()
```

```python
df_clean.to_csv('data/titanic_clean.csv', index=False)
print("Datos guardados en data/titanic_clean.csv")
```
